```python
import kaleido
import pandas as pd
import plotly as py
import plotly.express as px
import plotly.graph_objects as go

from dataclasses import dataclass
from pathlib import Path
from plotly.subplots import make_subplots
```

```python
%%capture
# Configure imported packages
py.offline.init_notebook_mode()
kaleido.get_chrome_sync()
```

```python
colormap = {
    'base': px.colors.qualitative.Plotly[0],
    'stm': px.colors.qualitative.Plotly[1],
    'unagi-chan': px.colors.qualitative.Plotly[2],
    'lossy-chan-2': px.colors.qualitative.Plotly[3],
    'lossy-chan-8': px.colors.qualitative.Plotly[4],
    'lossy-chan-32': px.colors.qualitative.Plotly[5],
    'lossy-chan-128': px.colors.qualitative.Plotly[6],
}
```

```python
df_time = pd.read_csv('lossy-chan-benchmark-time-2026-11-19.csv')
```

```python
fig = go.Figure()
for index, Name in enumerate(df_time.Name.unique()):
    fig.add_trace(go.Bar(
        name=Name,
        x=df_time.Benchmark.unique(),
        y=df_time[df_time.Name == Name].Mean,
        error_y=dict(type='data', array=df_time[df_time.Name == Name].Stddev),
        marker=dict(color=px.colors.qualitative.Plotly[index]),
        legendgroup=Name,
    ))
fig.update_layout(
    barmode='group',
    width=800,
    height=400,
)
fig.update_yaxes(title_text="Time (seconds)")
fig.show()
```

```python
df_space = pd.read_csv('lossy-chan-benchmark-space-2026-11-19.csv')
df_space.Allocated = df_space.Allocated.map(lambda v: v.replace(',','')).astype('int')
df_space.Max = df_space.Max.map(lambda v: v.replace(',','')).astype('int')
```

```python
fig = make_subplots(
        rows=3,
        cols=1,
        subplot_titles=(
            "Total Allocated",
            "Max Residency",
            "Max Residency, Zoomed In",
        ),
        vertical_spacing=0.1,
    )

for index, Name in enumerate(df_space.Name.unique()):
    fig.add_trace(go.Bar(
        name=Name,
        x=df_space.Benchmark.unique(),
        y=df_space[df_space.Name == Name].Allocated,
        marker=dict(color=px.colors.qualitative.Plotly[index]),
        legendgroup=Name,
    ),
    row=1,
    col=1,
    )
    fig.add_trace(go.Bar(
        name=Name,
        x=df_space.Benchmark.unique(),
        y=df_space[df_space.Name == Name].Max,
        marker=dict(color=px.colors.qualitative.Plotly[index]),
        showlegend=False,
        legendgroup=Name,
    ),
    row=2,
    col=1,
    )
    fig.add_trace(go.Bar(
        name=Name,
        x=df_space.Benchmark.unique(),
        y=df_space[df_space.Name == Name].Max,
        marker=dict(color=px.colors.qualitative.Plotly[index]),
        showlegend=False,
        legendgroup=Name,
    ),
    row=3,
    col=1,
    )
fig.update_layout(
    barmode='group',
    width=800,
    height=800,
)
fig.update_xaxes(
    showticklabels=False,
    row=1,
    col=1,
)
fig.update_xaxes(
    showticklabels=False,
    row=2,
    col=1,
)
fig.update_yaxes(
    range=[0,15000],
    row=3,
    col=1,
)
fig.show()
```
